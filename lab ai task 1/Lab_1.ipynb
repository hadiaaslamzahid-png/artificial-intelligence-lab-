```python
def symbolic(score, complete):
    return "Review" if complete and score >= 70 else "Hold"

def threshold(probability):
    return "Review" if probability >= 0.70 else "Hold"

cases = [
    ("complete_high", 82, True, 0.81),
    ("complete_low", 68, True, 0.74),
    ("incomplete_high", 91, False, 0.88),
    ("boundary", 70, True, 0.70)
]

print("case | symbolic | threshold | disagreement")
for name, score, complete, probability in cases:
    a, b = symbolic(score, complete), threshold(probability)
    print(name, a, b, a != b, sep=" | ")
```

```python
print("score 69:", symbolic(69, True), threshold(0.70))
print("probability 0.69:", symbolic(70, True), threshold(0.69))
```

```python
def check_application(score, complete, gpa, attendance):
    if not complete:
        return "Hold", "Documents are incomplete"
    if score < 70:
        return "Hold", "Score is below 70"
    if gpa < 2.0:
        return "Hold", "GPA is below 2.0"
    if attendance < 75:
        return "Hold", "Attendance is below 75"
    return "Review", "All conditions are met"

print("Thresholds are synthetic lab assumptions.")
cases = [
    ("pass", 80, True, 3.0, 85, "Review"),
    ("low_gpa", 85, True, 1.8, 90, "Hold"),
    ("low_attendance", 90, True, 3.2, 60, "Hold"),
    ("boundary", 70, True, 2.0, 75, "Review")
]
for name, score, complete, gpa, attendance, expected in cases:
    actual, reason = check_application(score, complete, gpa, attendance)
    print(f"{name} | input: {score}, {complete}, {gpa}, {attendance} | expected: {expected} | actual: {actual} | reason: {reason}")
```
