```python
def minimax(depth, index, max_turn, scores, target):
    if depth == target:
        return scores[index], 1
    left, count1 = minimax(depth + 1, index * 2, not max_turn, scores, target)
    right, count2 = minimax(depth + 1, index * 2 + 1, not max_turn, scores, target)
    value = max(left, right) if max_turn else min(left, right)
    return value, count1 + count2

def alphabeta(depth, index, max_turn, scores, target, alpha, beta):
    if depth == target:
        return scores[index], 1
    value = float('-inf') if max_turn else float('inf')
    leaves = 0
    for i in range(2):
        result, count = alphabeta(depth + 1, index * 2 + i, not max_turn, scores, target, alpha, beta)
        leaves += count
        if max_turn:
            value = max(value, result)
            alpha = max(alpha, value)
        else:
            value = min(value, result)
            beta = min(beta, value)
        if alpha >= beta:
            break
    return value, leaves

def compare(scores):
    target = len(scores).bit_length() - 1
    value1, leaves1 = minimax(0, 0, True, scores, target)
    value2, leaves2 = alphabeta(0, 0, True, scores, target, float('-inf'), float('inf'))
    print("Scores:", scores)
    print("Minimax value:", value1, "| Leaves:", leaves1)
    print("Alpha-Beta value:", value2, "| Leaves:", leaves2)
    assert value1 == value2
    assert leaves2 <= leaves1
```

```python
scores = [3, 5, 2, 9, 0, 7]
target = (len(scores) - 1).bit_length()
scores += [0] * (2 ** target - len(scores))
compare(scores)

scores = [3, 5, 2, 9, 3, 5, 2, 9]
compare(scores)
```
